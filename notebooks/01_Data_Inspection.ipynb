```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
import os

file_path = "../data/raw/train.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
print("Dataset Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
print("First 5 Records:")
display(df.head())

print("Last 5 Records:")
display(df.tail())
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Minimum Date:", df["date"].min())
print("Maximum Date:", df["date"].max())

print("\nUnique Stores:", df["store"].nunique())
print("Unique Items:", df["item"].nunique())

print("\nSales Summary:")
print(df["sales"].describe())
```

```python
print("Negative Sales:", (df["sales"] < 0).sum())
print("Zero Sales:", (df["sales"] == 0).sum())

print("Minimum Sales:", df["sales"].min())
print("Maximum Sales:", df["sales"].max())
```

```python
duplicate_keys = df.duplicated(
    subset=["date", "store", "item"]
).sum()

print("Duplicate Store-Item-Date Records:", duplicate_keys)
```

```python
parsed_dates = pd.to_datetime(
    df["date"],
    errors="coerce"
)

print("Invalid Dates:", parsed_dates.isna().sum())

print("Earliest Date:", parsed_dates.min())
print("Latest Date:", parsed_dates.max())
```

```python
print("Store IDs:")
print(sorted(df["store"].unique()))

print("\nItem ID Range:")
print(df["item"].min(), "to", df["item"].max())

print("\nInvalid Store IDs:", (df["store"] <= 0).sum())
print("Invalid Item IDs:", (df["item"] <= 0).sum())
```

total_days = (parsed_dates.max() - parsed_dates.min()).days + 1

expected_records = (
    total_days
    * df["store"].nunique()
    * df["item"].nunique()
)

print("Total Days:", total_days)
print("Expected Records:", expected_records)
print("Actual Records:", len(df))
print("Difference:", expected_records - len(df))

```python
expected_dates = pd.date_range(
    start=parsed_dates.min(),
    end=parsed_dates.max(),
    freq="D"
)

actual_dates = pd.DatetimeIndex(parsed_dates.unique())

missing_dates = expected_dates.difference(actual_dates)

print("Missing Calendar Dates:", len(missing_dates))

store_item_counts = df.groupby(
    ["store", "item"]
)["date"].nunique()

print("Total Store-Item Combinations:", len(store_item_counts))

print(
    "Combinations with Incomplete Date Coverage:",
    (store_item_counts != len(expected_dates)).sum()
)
```
