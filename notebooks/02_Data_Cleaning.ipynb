# Sales Forecasting & Demand Prediction Analysis

## Notebook 02 — Data Cleaning & Transformation

### Objective
Prepare the historical sales dataset for exploratory
data analysis and demand forecasting.

### Planned Tasks
1. Load the original dataset.
2. Convert the date column to datetime.
3. Validate data types and sales quantities.
4. Create time-based features.
5. Prepare clean data for analysis.
6. Export the processed dataset.

### Data Integrity
The original dataset in data/raw/ will remain unchanged.

```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries imported successfully!")
```

```python
file_path = Path("../data/raw/train.csv")

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
df_clean = df.copy()

print("Working copy created successfully!")
print("Original Shape:", df.shape)
print("Working Copy Shape:", df_clean.shape)
```

```python
print("Data Types Before Conversion:")
print(df_clean.dtypes)
```

```python
df_clean["date"] = pd.to_datetime(
    df_clean["date"],
    format="%Y-%m-%d",
    errors="raise"
)

print("Date conversion completed successfully!")
```

```python
print("Data Types After Conversion:")
print(df_clean.dtypes)

print("\nFirst 5 Records:")
display(df_clean.head())
```

```python
print("Earliest Date:", df_clean["date"].min())
print("Latest Date:", df_clean["date"].max())

print("\nMissing Values After Conversion:")
print(df_clean.isnull().sum())
```

```python
print("Original Records:", len(df))
print("Cleaned Records:", len(df_clean))

print("\nDate Column Is Datetime:",
      pd.api.types.is_datetime64_any_dtype(df_clean["date"]))

print("Store Is Numeric:",
      pd.api.types.is_numeric_dtype(df_clean["store"]))

print("Item Is Numeric:",
      pd.api.types.is_numeric_dtype(df_clean["item"]))

print("Sales Is Numeric:",
      pd.api.types.is_numeric_dtype(df_clean["sales"]))

print("\nMissing Values:", df_clean.isnull().sum().sum())
```

```python
df_clean["Year"] = df_clean["date"].dt.year

print("Year column created successfully!")
print(df_clean["Year"].unique())
```

```python
df_clean["Month"] = df_clean["date"].dt.month
df_clean["Quarter"] = df_clean["date"].dt.quarter

print("Month and Quarter columns created!")

display(
    df_clean[["date", "Month", "Quarter"]].head()
)
```

```python
df_clean["Day"] = df_clean["date"].dt.day

df_clean["Day_of_Week"] = df_clean["date"].dt.day_of_week

print("Day and Day_of_Week columns created!")

display(
    df_clean[["date", "Day", "Day_of_Week"]].head()
)
```

```python
df_clean["Is_Weekend"] = (
    df_clean["Day_of_Week"] >= 5
).astype(int)

print("Weekend indicator created!")

display(
    df_clean[
        ["date", "Day_of_Week", "Is_Weekend"]
    ].head(10)
)
```

```python
print("Dataset Shape:", df_clean.shape)

print("\nColumn Names:")
print(df_clean.columns.tolist())

print("\nData Types:")
print(df_clean.dtypes)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nFirst 5 Records:")
display(df_clean.head())
```

```python
# Step 4.4.1 - Final Data Validation

print("========== FINAL DATA VALIDATION ==========")

# 1. Check dataset shape
print("Dataset Shape:", df_clean.shape)

# 2. Check missing values
print("Total Missing Values:",
      df_clean.isnull().sum().sum())

# 3. Check duplicate business keys
print("Duplicate Store-Item-Date Records:",
      df_clean.duplicated(
          subset=["date", "store", "item"]
      ).sum())

# 4. Check negative sales
print("Negative Sales:",
      (df_clean["sales"] < 0).sum())

# 5. Check zero sales
print("Zero Sales:",
      (df_clean["sales"] == 0).sum())

# 6. Check date range
print("Earliest Date:", df_clean["date"].min())
print("Latest Date:", df_clean["date"].max())

# 7. Check unique stores and items
print("Unique Stores:", df_clean["store"].nunique())
print("Unique Items:", df_clean["item"].nunique())

# 8. Verify weekend indicator
expected_weekend = (
    df_clean["date"].dt.day_of_week >= 5
).astype(int)

print("Incorrect Weekend Indicators:",
      (df_clean["Is_Weekend"] != expected_weekend).sum())

print("========== VALIDATION COMPLETE ==========")
```

```python
from pathlib import Path

# Define output location
output_path = Path("../data/processed/sales_cleaned.csv")

# Ensure output folder exists
output_path.parent.mkdir(parents=True, exist_ok=True)

# Export cleaned dataset
df_clean.to_csv(output_path, index=False)

print("Cleaned dataset exported successfully!")
print("Saved Location:", output_path.resolve())
```
